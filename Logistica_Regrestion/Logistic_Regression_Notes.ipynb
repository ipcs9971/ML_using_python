# 1. What is Logistic Regression?

Logistic Regression is a supervised machine-learning algorithm used mainly for classification problems.

Unlike Linear Regression, which predicts a continuous number:

House price = ₹50,00,000

Logistic Regression predicts a class/probability:

Will the customer buy? → Yes/No
Is the patient diabetic? → Yes/No
Is the email spam? → Spam/Not Spam

Although its name contains "Regression", it is primarily used for classification.

Simple example

Suppose we want to predict whether a student will pass:

Hours Studied	Result
1	Fail
2	Fail
3	Fail
5	Pass
6	Pass
8	Pass

Our model learns the relationship between:

Hours Studied → Pass/Fail


2. Linear Regression vs Logistic Regression
Linear Regression	Logistic Regression
Regression algorithm	Classification algorithm
Predicts continuous values	Predicts classes/probabilities
Output can be any number	Output probability between 0 and 1
Example: salary prediction	Example: pass/fail
Uses linear equation	Uses sigmoid function

Linear Regression:

y = b0 + b1x

Logistic Regression first calculates:

z = b0 + b1x

Then applies the Sigmoid function:

P(y=1) = 1 / (1 + e^-z)

The sigmoid converts the result into a probability between 0 and 1.

3. How Logistic Regression works

Imagine the model produces:

0.10
0.25
0.48
0.72
0.91

These can be interpreted as probabilities of class 1.

Usually:

Probability >= 0.5 → Class 1
Probability < 0.5  → Class 0

For example:

0.82 → 1 → Yes
0.31 → 0 → No

The threshold does not always have to be 0.5; it can be adjusted depending on the problem.

4. Types of Logistic Regression

There are three important types.

Type 1 — Binary Logistic Regression

Used when there are two classes.

Yes / No
Pass / Fail
0 / 1
Disease / No Disease
Spam / Not Spam

Example:

from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
Type 2 — Multinomial Logistic Regression

Used when there are more than two classes and they are not naturally ordered.

Example:

Car
Bike
Bus

or:

Setosa
Versicolor
Virginica

Python:

model = LogisticRegression(
    multi_class="multinomial",
    max_iter=1000
)

Note: In newer versions of scikit-learn, some solver/multiclass behavior is handled automatically, so you will often simply use:

model = LogisticRegression(max_iter=1000)
Type 3 — Ordinal Logistic Regression

Used when categories have a meaningful order.

Example:

Poor
Average
Good
Excellent

The classes have an order:

Poor < Average < Good < Excellent

Standard sklearn.linear_model.LogisticRegression is not specifically an ordinal logistic-regression implementation. Specialized statistical libraries/models are normally used for ordinal logistic regression.

5. Important Logistic Regression parameters
LogisticRegression(
    penalty="l2",
    C=1.0,
    solver="lbfgs",
    max_iter=1000
)
penalty

Controls regularization.

Common choices:

l1
l2
elasticnet
None
C

Controls the strength of regularization.

Important:

Small C → stronger regularization
Large C → weaker regularization
solver

Optimization algorithm.

Common solvers include:

lbfgs
liblinear
saga
newton-cg
newton-cholesky
max_iter

Maximum number of iterations.

If you get:

ConvergenceWarning

try:

max_iter=1000

8. Example 3 — Diabetes Prediction

This is a common classification problem.

Suppose we have:

age
bmi
blood_glucose

Target:

diabetes
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

data = {
    "age": [25, 30, 35, 40, 45, 50, 55, 60],
    "bmi": [22, 25, 27, 30, 32, 35, 36, 40],
    "glucose": [90, 100, 110, 130, 145, 160, 180, 200],
    "diabetes": [0, 0, 0, 0, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

X = df[["age", "bmi", "glucose"]]
y = df["diabetes"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

model = LogisticRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)

For a real diabetes project, you would need a sufficiently large dataset and proper validation rather than relying on this tiny demonstration dataset.

A simple demonstration using word counts:

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression

messages = [
    "win free money",
    "claim your prize",
    "free lottery winner",
    "hello how are you",
    "meeting at 5 pm",
    "please call me"
]

y = [1, 1, 1, 0, 0, 0]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(messages)

model = LogisticRegression()

model.fit(X, y)

new_message = ["free money winner"]

X_new = vectorizer.transform(new_message)

prediction = model.predict(X_new)

print(prediction)

This demonstrates the basic idea:

Text
 ↓
Convert text to numbers
 ↓
Logistic Regression
 ↓
Spam / Not Spam

For real text classification, TF-IDF is often a stronger baseline than raw word counts.

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

    "age": [20, 22, 25, 28, 30, 35, 40, 45, 50, 55],

38. Cross Validation

A single train/test split can sometimes give a misleading result.

Cross-validation gives a more reliable estimate.

Example:

scores = cross_val_score(
    model,
    X_train_scaled,
    y_train,
    cv=5,
    scoring="accuracy"
)

print("CV Scores:", scores)

print(
    "Mean CV Accuracy:",
    scores.mean()
)

With 5-fold cross-validation:

Fold 1
Fold 2
Fold 3
Fold 4
Fold 5

Each fold gets a chance to act as the validation portion.

44. Better Approach: Use a Pipeline

In real projects, I strongly recommend using a Pipeline.

Instead of manually doing:

Scaler
 ↓
Logistic Regression

you can combine them.

from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000))
])

Train:

pipeline.fit(
    X_train,
    y_train
)

Predict:

y_pred = pipeline.predict(
    X_test
)

This is safer and cleaner.

45. Pipeline + GridSearchCV

Even better:

pipeline = Pipeline([

    ("model", LogisticRegression(max_iter=1000))

Parameter grid:

model__C

24. Data Leakage

This is one of the most important ML concepts.

Suppose you calculate the scaler using the entire dataset:

before splitting.

Then information from the test data influences preprocessing.

That's leakage.

Correct:

Split
 ↓
Fit preprocessing on training data
 ↓
Transform training data
 ↓
Transform test data using training preprocessing

Even better:

Pipeline

Use appropriate metrics:

Mistake 3

Testing many models on the test set and choosing the winner.

Encoding before splitting without considering leakage.
